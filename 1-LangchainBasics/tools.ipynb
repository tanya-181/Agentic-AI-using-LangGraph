```python
## @tool decorator

from langchain_core.tools import tool

@tool
def divison(a:int, b:int) -> int:
    """Divides two numbers"""
    return a / b
```

```python
print(divison.name)
print(divison.description)
print(divison.args)
```

```python
#async implementation

@tool
async def async_division(a: int, b: int) -> int:
    """Divides two numbers asynchronously"""
    return a / b
```

```python
from typing import Annotated, List

@tool
def multiply_by_max(
    a: Annotated[int, "A value"],
    b: Annotated[List[int], "B list value"]
) -> int:
    """Multiply a by the maximum b"""
    return a * max(b)
```

```python
print(multiply_by_max.args)
```

### Structured tool


```python
from langchain_core.tools import StructuredTool

def multiply(a:int, b:int) -> int:
    """Multiply two numbers"""
    return a * b

async def amultiply(a:int, b:int) -> int:
    """Multiply two numbers asynchronously"""
    return a * b

calculator = StructuredTool.from_function(func=multiply, coroutine=amultiply)

print(calculator.invoke({"a":2, "b":5}))
print(await calculator.ainvoke({"a":2, "b":5}))
```

CALL Tool with LLM MOdel

```python
from langchain_core.tools import tool

@tool
def add(a:int, b:int) -> int:
    """Adds two numbers"""
    return a + b

@tool
def multiply(a:int, b:int) -> int:
    """Multiply two numbers"""
    return a * b
```

```python
tools=[wiki_tool, add, multiply, tavily_tool]
```

```python
tools
```

```python
from langchain.chat_models import init_chat_model
llm = init_chat_model("openai/gpt-oss-20b", model_provider="groq")
```

```python
llm_with_tool = llm.bind_tools(tools)
llm_with_tool
```

```python
from langchain_core.messages import HumanMessage
query = "What is 2 multiply 3"
messages = [HumanMessage(query)]
response = llm_with_tool.invoke(query)
print(response)
```

```python
response
```

```python
response.tool_calls
```

```python
for tool_call in response.tool_calls:
    print(tool_call["name"])
```

```python
add.name
```

```python
multiply.name
```

```python
tavily_tool.name
```

```python
for tool_call in response.tool_calls:
    selected_tool= {"add": add, "multiply": multiply, "wiki_tool": wiki_tool, "tavily_search": tavily_tool}[tool_call["name"].lower()]
    tool_msg = selected_tool.invoke(tool_call)
    messages.append(tool_msg)

messages    
```

```python
response = llm_with_tool.invoke(messages)
```
